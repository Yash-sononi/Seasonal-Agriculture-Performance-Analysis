```python
import pandas as pd

# Load the uploaded dataset
df = pd.read_csv("seasonal_agriculture_performance_dataset (2) (1).csv")

# Show first few rows
df.head()
```

```python
# Check dataset structure
df.info()

# View summary statistics
df.describe()
```

```python
# Data Cleaning and Preparation

# Standardize season names
df['Season'] = df['Season'].str.strip().str.lower()

# Convert numeric columns to proper types
numeric_cols = ['Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Sunlight_Hours_Day']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

print("Data cleaned successfully!")
```

```python
# Seasonal Analysis
season_summary = df.groupby('Season')[numeric_cols].mean()
print(season_summary)
```

```python
import matplotlib.pyplot as plt

season_summary.plot(kind='bar', figsize=(10,6))
plt.title("Seasonal Averages of Climate Factors")
plt.xlabel("Season")
plt.ylabel("Average Values")
plt.legend(title="Factors")
plt.show()
```

The seasonal analysis shows clear climatic variations across agricultural seasons.
Rainfall is highest during the kharif season (around 850 mm), supporting crops like rice and maize.
Humidity also peaks in kharif, while sunlight hours are relatively lower due to cloud cover.
In the rabi season, rainfall decreases to nearly half, temperature drops moderately, and humidity reduces, creating ideal conditions for wheat and pulses.
The zaid season experiences the least rainfall but the longest sunlight hours, favoring short‑duration crops like cucumber and watermelon.
Overall, kharif is the wettest, rabi is the coolest, and zaid is the driest and sunniest season.


Conclusion:
The seasonal analysis confirms that climatic variations strongly influence agricultural performance.
Kharif season, with high rainfall and humidity, supports water‑intensive crops like rice and maize.
Rabi season, with moderate rainfall and cooler temperatures, favors wheat and pulses.
Zaid season, with minimal rainfall but maximum sunlight, is suitable for short‑duration crops such as cucumber and watermelon.
These insights highlight how farmers can plan crop selection and resource allocation based on seasonal conditions.


```python
# Performance Insights
performance_summary = df.groupby('Season')[['Yield_Tonnes_Ha','Profit_INR']].mean()
print(performance_summary)
```

Performance Insights:
Kharif season shows high yields but also higher costs due to irrigation and pest risks.
Rabi season balances moderate yields with stable profits, making it economically reliable.
Zaid season has lower yields but can be profitable for short‑duration, high‑market‑value crops.
Overall, seasonal planning helps maximize both productivity and profitability.
